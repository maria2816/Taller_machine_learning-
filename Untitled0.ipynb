## Model Comparisan

1.In this project, we trained and evaluated a Decision Tree Classifier and a
K-Nearest Neighbors (k-NN) model on the Iris dataset.
2.The Decision Tree model offers high interpretability because its logical branches can be easily visualized and understood by humans.
3.On the other hand, the k-NN model relies on distance metrics between samples,making it less interpretable as a direct rule-based system.
4.In terms of operational risk, k-NN carries a higher computational risk when scaled to very large datasets because it requires storing and scanning all training instances.
5.Ultimately, both models achieved excellent metrics due to the low complexity of the Iris dataset, but the Decision tree is prederred if structural transparency is required.
